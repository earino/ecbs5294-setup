# Setup check — the notebook half

`check.py` proved that `uv sync` and `uv run` work in the terminal. This notebook proves that VS Code can **run
cells** on this project's environment, and that DuckDB answers a query there. Every lab starts this way.

1. Click **Select Kernel** (top right) → **Python Environments…** → the entry marked *Recommended* whose path
   contains **`.venv`**.
2. **Run All** (the ▷▷ button above, or *Run → Run All*).
3. Three outputs appear: a path that contains `.venv`, the line `duckdb 1.5.5`, and a small table with one
   number in it, **525461**.
4. **Save the notebook** (Ctrl+S, or Cmd+S on a Mac), so that the outputs are in the file you commit in README
   step 4.

```python
import sys
print(sys.executable)   # must contain .venv; otherwise you picked another Python
```

```python
import duckdb
print("duckdb", duckdb.__version__)   # must be 1.5.5; ModuleNotFoundError here means the wrong kernel
```

```python
con = duckdb.connect()   # an in-memory database: it reads files directly, with no loading step
con.sql("SELECT COUNT(*) AS invoice_lines FROM 'data/raw/online_retail.parquet'").df()   # 525461: one row per invoice line
```

**If something is wrong:**

- The first line does not contain `.venv`, or the second cell fails: open the kernel picker again and choose the
  `.venv` entry. A global Python, Anaconda, or another project's kernel is a different environment.
- The third cell says `IO Error: No files found`: the notebook is not running from this folder. Open the
  **folder** in VS Code (*File → Open Folder…*), not the notebook file on its own.
